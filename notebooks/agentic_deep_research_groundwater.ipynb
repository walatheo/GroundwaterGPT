This notebook demonstrates the implementation of agentic deep research techniques for groundwater research, using priority ranking and intelligent searching as described in **Awesome-Deep-Research** (https://github.com/DavidZWZ/Awesome-Deep-Research).

## Overview

We implement the following research patterns:
- **SmartSearch**: Priority ranking of research results by relevance, trust, and recency
- **ReSeek**: Budget-aware knowledge-seeking with query optimization
- **WebSeer**: Self-reflection for quality assessment and gap identification
- **O-Researcher**: Query decomposition into sub-research tasks

## Research Process

1. **Research Planning**: Decompose complex groundwater questions into sub-questions
2. **Priority Ranking**: Order search queries by research importance
3. **Multi-Source Searching**: Query knowledge base, web, and USGS data
4. **Insight Extraction**: Extract and verify findings from search results
5. **Synthesis**: Build structured report with citations
6. **Self-Reflection**: Evaluate quality and identify gaps
7. **Iteration**: Continue searching if quality threshold not met

## 6. Develop Groundwater Model with Agent-Driven Insights

The groundwater model incorporates:
- Domain knowledge (aquifer properties)
- Data validation rules
- Pattern detection (seasonal, anomalies)
- Multi-site analysis
- Agent-extracted insights

```python
# Create synthetic groundwater data for Biscayne Aquifer
print("📊 SYNTHETIC BISCAYNE AQUIFER DATA GENERATION\n")

np.random.seed(42)
dates = pd.date_range('2020-01-01', periods=730, freq='D')  # 2 years
months = dates.month

# Biscayne Aquifer: seasonal pattern with tides
base_signal = 5.0 + 1.5 * np.sin(2 * np.pi * months / 12)  # 1.5 ft amplitude
noise = np.random.normal(0, 0.3, len(dates))  # Add measurement noise
tidal_signal = 0.5 * np.sin(2 * np.pi * np.arange(len(dates)) / 14)  # ~14-day tidal cycles

elevation = base_signal + noise + tidal_signal

# Add an anomaly (spike)
elevation[200] = 12.0  # Spike

biscayne_data = pd.DataFrame({
    'date': dates,
    'elevation_ft': elevation,
    'water_level_ft_below_surface': 35 - elevation,  # ~35 ft to water table
})

print(f"Generated synthetic data:")
print(f"  - Period: {biscayne_data['date'].min().date()} to {biscayne_data['date'].max().date()}")
print(f"  - Records: {len(biscayne_data)}")
print(f"  - Mean elevation: {elevation.mean():.2f} ft")
print(f"  - Std dev: {elevation.std():.2f} ft")
print(f"\nData sample:")
print(biscayne_data.head(10))
```

```python
# Detect seasonal patterns
print("\n🌡️  SEASONAL PATTERN DETECTION\n")

pattern = model.detect_seasonal_pattern(biscayne_data, AquiferType.BISCAYNE, "USG_251241080385301")

if pattern:
    print(f"✅ Seasonal pattern detected:")
    print(f"   Peak month: {pattern.peak_month} (elevation: {pattern.peak_elevation:.2f} ft)")
    print(f"   Trough month: {pattern.trough_month} (elevation: {pattern.trough_elevation:.2f} ft)")
    print(f"   Annual amplitude: {pattern.annual_amplitude:.2f} ft")
    print(f"   Predictability score: {pattern.predictability_score:.2f}/1.0")
    print(f"\n   Interpretation: {pattern.interpret()}")
else:
    print("No seasonal pattern detected (insufficient data span)")

# Detect anomalies
print("\n\n🚨 ANOMALY DETECTION\n")

anomalies = model.detect_anomalies(biscayne_data, AquiferType.BISCAYNE, "USG_251241080385301", threshold=3.0)

print(f"Found {len(anomalies)} anomalies:")
for i, anomaly in enumerate(anomalies, 1):
    print(f"\n   {i}. {anomaly.interpret()}")
    print(f"      Severity: {anomaly.severity:.2f}/1.0")
```

```python
# Visualization of water level time series
fig, axes = plt.subplots(2, 1, figsize=(14, 8))

# Plot 1: Full time series
ax = axes[0]
ax.plot(biscayne_data['date'], biscayne_data['elevation_ft'], label='Water Elevation', linewidth=1.5, alpha=0.7)
if anomalies:
    anomaly_dates = [a.date for a in anomalies]
    anomaly_values = [a.value for a in anomalies]
    ax.scatter(anomaly_dates, anomaly_values, color='red', s=100, marker='o', label='Detected Anomalies', zorder=5)
ax.set_xlabel('Date')
ax.set_ylabel('Elevation (ft MSL)')
ax.set_title('Biscayne Aquifer Water Level - 2 Year Time Series')
ax.legend()
ax.grid(True, alpha=0.3)

# Plot 2: Monthly average (seasonal pattern)
ax = axes[1]
monthly_mean = biscayne_data.groupby(biscayne_data['date'].dt.month)['elevation_ft'].mean()
months_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
ax.plot(range(1, 13), monthly_mean.values, marker='o', linewidth=2, markersize=8, label='Monthly Average')
ax.fill_between(range(1, 13), monthly_mean.values, alpha=0.3)
if pattern:
    ax.axvline(pattern.peak_month, color='green', linestyle='--', label=f'Peak (Month {pattern.peak_month})')
    ax.axvline(pattern.trough_month, color='red', linestyle='--', label=f'Trough (Month {pattern.trough_month})')
ax.set_xlabel('Month')
ax.set_ylabel('Mean Elevation (ft MSL)')
ax.set_title('Seasonal Pattern - Monthly Averages')
ax.set_xticks(range(1, 13))
ax.set_xticklabels(months_names, rotation=45)
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("✅ Visualization complete")
```
